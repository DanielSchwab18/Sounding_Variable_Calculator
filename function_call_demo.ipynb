```python
s = bindings.Sounding("example", False, False, 100)

print("Surface-based CAPE (sb3cape):", s.get_sb3cape())
print("Surface-based ECAPE (sb3ecape):", s.get_sb3ecape())

print("Mixed-layer CAPE (ml3cape):", s.get_ml3cape())
print("Mixed-layer ECAPE (ml3ecape):", s.get_ml3ecape())

print("Most-unstable CAPE (mu3cape):", s.get_mu3cape())
print("Most-unstable ECAPE (mu3ecape):", s.get_mu3ecape())

print("SB Parcel Buoyancy Trace: ", s.get_sb_parcel_buoyancy())
```
